```python
#monalika_kubde-23070521094

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Bidirectional
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

sentences = [
    "book a flight from mumbai to delhi",
    "i want a flight from delhi to bangalore",
    "show flights from pune to mumbai",
    "find a flight from chennai to delhi",
    "book a ticket from nagpur to goa",
    "i need a flight from hyderabad to pune",
    "find flights from bangalore to chennai",
    "book flight from kolkata to mumbai",
    "show me flights from delhi to goa",
    "i want to travel from pune to delhi",
    "book a flight from jaipur to mumbai",
    "find a flight from surat to delhi",
    "show flights from goa to pune",
    "i need a ticket from mumbai to chennai",
    "book a flight from delhi to hyderabad",
    "find flights from nagpur to bangalore",
    "i want a ticket from pune to kolkata",
    "book flight from chennai to mumbai",
    "show me flights from goa to delhi",
    "find a flight from hyderabad to bangalore"
]

labels = [
    ["O","O","O","O","B-fromloc.city","O","B-toloc.city"],
    ["O","O","O","O","B-fromloc.city","O","B-toloc.city"],
    ["O","O","O","O","B-fromloc.city","O","B-toloc.city"],
    ["O","O","O","O","B-fromloc.city","O","B-toloc.city"],
    ["O","O","O","O","B-fromloc.city","O","B-toloc.city"],
    ["O","O","O","O","B-fromloc.city","O","B-toloc.city"],
    ["O","O","O","O","B-fromloc.city","O","B-toloc.city"],
    ["O","O","O","O","B-fromloc.city","O","B-toloc.city"],
    ["O","O","O","O","O","B-fromloc.city","O","B-toloc.city"],
    ["O","O","O","O","B-fromloc.city","O","B-toloc.city"],
    ["O","O","O","O","B-fromloc.city","O","B-toloc.city"],
    ["O","O","O","O","B-fromloc.city","O","B-toloc.city"],
    ["O","O","O","O","B-fromloc.city","O","B-toloc.city"],
    ["O","O","O","O","B-fromloc.city","O","B-toloc.city"],
    ["O","O","O","O","B-fromloc.city","O","B-toloc.city"],
    ["O","O","O","O","B-fromloc.city","O","B-toloc.city"],
    ["O","O","O","O","B-fromloc.city","O","B-toloc.city"],
    ["O","O","O","O","B-fromloc.city","O","B-toloc.city"],
    ["O","O","O","O","O","B-fromloc.city","O","B-toloc.city"],
    ["O","O","O","O","B-fromloc.city","O","B-toloc.city"]
]

print("\n======================================")
print("TEXT PREPROCESSING")
print("======================================")

# Create tokenizer
tokenizer = Tokenizer(
    lower=True,
    oov_token="<OOV>"
)

tokenizer.fit_on_texts(sentences)

# Word dictionary
word_index = tokenizer.word_index

print("\nWord Index:")
print(word_index)

# Convert sentences into numbers
X = tokenizer.texts_to_sequences(sentences)

# Maximum sentence length
MAX_LEN = max(len(x) for x in X)

print("\nMaximum sentence length:", MAX_LEN)

# Padding
X = pad_sequences(
    X,
    maxlen=MAX_LEN,
    padding="post"
)

print("\nPreprocessed Sentences:")
print(X)

print("\n======================================")
print("LABEL ENCODING")
print("======================================")

# Find unique labels
unique_labels = sorted(
    set(label for sequence in labels for label in sequence)
)

# Create label-to-ID mapping
label2id = {
    label: i + 1
    for i, label in enumerate(unique_labels)
}

# 0 is reserved for padding
label2id["PAD"] = 0

# Reverse mapping
id2label = {
    value: key
    for key, value in label2id.items()
}

print("\nLabel Mapping:")
print(label2id)


# Convert labels into numbers
Y = []

for sequence in labels:

    encoded = [
        label2id[label]
        for label in sequence
    ]

    # Padding
    while len(encoded) < MAX_LEN:
        encoded.append(label2id["PAD"])

    Y.append(encoded)

Y = np.array(Y)

print("\nEncoded Labels:")
print(Y)

X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.20,
    random_state=42
)

print("\n======================================")
print("DATASET SPLIT")
print("======================================")

print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))

print("\n======================================")
print("BUILDING LSTM MODEL")
print("======================================")

VOCAB_SIZE = len(word_index) + 1
EMBEDDING_DIM = 64
LSTM_UNITS = 64
NUM_CLASSES = len(label2id)

model = Sequential()

# Embedding layer
model.add(
    Embedding(
        input_dim=VOCAB_SIZE,
        output_dim=EMBEDDING_DIM
    )
)

# Bidirectional LSTM
model.add(
    Bidirectional(
        LSTM(
            LSTM_UNITS,
            return_sequences=True
        )
    )
)

# Output layer
model.add(
    Dense(
        NUM_CLASSES,
        activation="softmax"
    )
)

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("\nModel Summary:")
model.summary()

print("\n======================================")
print("TRAINING MODEL")
print("======================================")

history = model.fit(
    X_train,
    Y_train,
    epochs=30,
    batch_size=2,
    validation_split=0.2,
    verbose=1
)

print("\n======================================")
print("MODEL EVALUATION")
print("======================================")

loss, accuracy = model.evaluate(
    X_test,
    Y_test,
    verbose=0
)

print("\nTest Loss     :", loss)
print("Test Accuracy :", accuracy)

Y_pred = model.predict(
    X_test,
    verbose=0
)

# Convert probabilities to class IDs
Y_pred = np.argmax(
    Y_pred,
    axis=-1
)

true_labels = []
pred_labels = []

for true_sequence, pred_sequence in zip(
    Y_test,
    Y_pred
):

    for true_id, pred_id in zip(
        true_sequence,
        pred_sequence
    ):

        # Ignore padding
        if true_id != label2id["PAD"]:

            true_labels.append(
                id2label[true_id]
            )

            pred_labels.append(
                id2label.get(
                    pred_id,
                    "O"
                )
            )


token_accuracy = accuracy_score(
    true_labels,
    pred_labels
)

print("\nToken Level Accuracy:",
      token_accuracy)

print("\n======================================")
print("CLASSIFICATION REPORT")
print("======================================")

print(
    classification_report(
        true_labels,
        pred_labels,
        zero_division=0
    )
)

plt.figure(figsize=(9, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy",
    linewidth=2,
    marker="o"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy",
    linewidth=2,
    marker="o"
)

plt.title(
    "LSTM Slot Filling - Training and Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

plt.figure(figsize=(9, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss",
    linewidth=2,
    marker="o"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss",
    linewidth=2,
    marker="o"
)

plt.title(
    "LSTM Slot Filling - Training and Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

def predict_slots(sentence):

    print("\n======================================")
    print("SLOT PREDICTION")
    print("======================================")

    print("Sentence:", sentence)

    # Convert sentence to numbers
    sequence = tokenizer.texts_to_sequences(
        [sentence]
    )

    # Padding
    padded = pad_sequences(
        sequence,
        maxlen=MAX_LEN,
        padding="post"
    )

    # Predict
    prediction = model.predict(
        padded,
        verbose=0
    )

    # Select highest probability
    prediction = np.argmax(
        prediction,
        axis=-1
    )[0]

    # Split sentence into words
    words = sentence.lower().split()

    print("\nWord                 Slot")
    print("--------------------------------")

    for i, word in enumerate(words):

        if i < MAX_LEN:

            slot = id2label.get(
                prediction[i],
                "O"
            )

            print(
                f"{word:<20} {slot}"
            )

predict_slots(
    "book a flight from mumbai to delhi"
)

predict_slots(
    "find a flight from pune to goa"
)

predict_slots(
    "book a flight from nagpur to chennai"
)
```
